```python
import torch
from torch.utils.data import Dataset
import os
import cv2
import numpy as np
```

```python
class CamVidDataset(Dataset):
    def __init__(self, images_dir, masks_dir):
        # Speichert die Pfade und listet alle verfügbaren Bilder auf
        self.images_dir = images_dir
        self.masks_dir = masks_dir
    
        # Liste aller Dateinamen im Bildordner
        self.images = os.listdir(images_dir)

    def __len__(self):
        # Sagt PyTorch, wie viele Bilder insgesamt im Datensatz existieren.
        return len(self.images)

    def __getitem__(self. idx):
        # Lädt ein einzelnes Bild und die dazugehörige Maske anhand eines Index (idx)
        img_name = self.images[idx]
        img_path = os.path.join(self.images, img_name)

        # Maskennamen anpassen und Suffix für CamVid hinzufügen
        mask_name = img_name.replace('.png', '_L.png')
        mask_path = os.path.join(self.masks_dir, mask_name)

        # 1. Bild laden und von BGR (OpenCV Standard) zu RGB konvertieren
        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        # 2. Maske/Label laden
        mask = cv2.imread(mask_path)
        mask = cv2.cvtColor(mask, cv2.COLOR_BGR2GRAY)

        # 3. In Tensoren umwandeln
        # PyTorch erwartet Bilder im Format (Channels, Height, Width), OpenCV liefert (Height, Width, Channels)
        image = torch.tensor(image, dtype=torch.float32).permute(2, 0, 1) / 255.0
        mask = torch.tensor(mask, dtype = torch.long)

        return image, mask
```

```python
train_dataset = CamVidDataset(images_dir="../data/train", masks_dir="../data/train_labels")

test_image, test_mask = train_dataset[0]

print(f"Anzahl der Bilder im Datensatz: {len(train_dataset)}")
print(f"Shape des Bild-Tensors: {test_image.shape}")
print(f"Shape des Masken-Tensors: {test_mask.shape}")
```
